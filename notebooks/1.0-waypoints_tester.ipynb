```python
import os
import random
import yaml
import numpy as np
```

```python
filename = './../scripts/waypoints.yaml'

n_points = 5
min_x = -1
min_y = -1
max_x = 1
max_y = 1
min_th = 0
max_th = 2*np.pi

res = 100

data = {}
```

# Writing points in wayponits.yaml

```python
for idx in range(n_points):
    x_ = random.randint(min_x*res, max_x*res)/res
    y_ = random.randint(min_y*res, max_y*res)/res
    thetha_ = random.randint(int(min_th*res), int(max_th*res))/res

    print(f'[x: {x_}, y: {y_}, th: {thetha_}]')

    data[idx] = [x_, y_, thetha_]
```

```python
with open(filename, 'w') as f:
    yaml.dump(data, f)

print(f"Waypoints file has been successfully written.")
```

# Reading waypoints.yaml

```python
with open(filename, 'r') as f:
    loaded_data = yaml.safe_load(f)

print(f"Data read from {filename}")
print(loaded_data)
```

```python
waypoints = list(loaded_data.values())
print(waypoints)
```

```python
random.shuffle(waypoints)
```
